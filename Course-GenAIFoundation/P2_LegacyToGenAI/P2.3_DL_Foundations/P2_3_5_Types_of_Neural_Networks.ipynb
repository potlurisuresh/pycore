## 🗺️ The Big Picture — 4 Types of Neural Networks

Think of neural networks like tools in a toolbox.
You pick the right architecture for the right problem.

| Architecture | Full Name | Best For | Example |
|---|---|---|---|
| **ANN** | Artificial Neural Network | Tabular / structured data | House price prediction |
| **CNN** | Convolutional Neural Network | Images, spatial data | Tumour detection from X-ray |
| **RNN** | Recurrent Neural Network | Short sequences, simple text | Phone keyboard next-word |
| **LSTM** | Long Short-Term Memory | Long sequences, speech-to-text | English → French translation |

The full code + training for each comes in the dedicated notebooks ahead.

## 🗺️ The Full Decision Table

Use this to pick the right network before you start any project:

```
────────────────────────────────────────────────────────────────────────────────
Data Type      Task                    Data Size       →  Approach
────────────────────────────────────────────────────────────────────────────────
Tabular        Regression              Any             →  ANN, train from scratch
Tabular        Binary Classification   Any             →  ANN, sigmoid output
Tabular        Multi Classification    Any             →  ANN, softmax output

Image          Classification          Large           →  CNN from scratch
Image          Classification          Small           →  Pretrained CNN (Transfer)
Image          Object Detection        Any             →  Pretrained CNN

Text           Sentiment Analysis      Any             →  RNN / LSTM
Text           Next Word Prediction    Any             →  RNN / LSTM
Audio          Speech to Text          Large           →  RNN / LSTM from scratch
Audio          Speech to Text          Small           →  Pretrained model
Time Series    Next Value Prediction   Any             →  RNN / LSTM
────────────────────────────────────────────────────────────────────────────────
```
